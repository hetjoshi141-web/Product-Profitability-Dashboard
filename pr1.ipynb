```python
# 1. Revenue by Product
product_revenue = (
    data1.groupby('Product Name')['Sales']
    .sum()
    .reset_index()
)

# 2. Sort by Sales (Highest to Lowest)
product_revenue = product_revenue.sort_values(
    by='Sales',
    ascending=False
)

# 3. Calculate Cumulative Sales
product_revenue['Cumulative Sales'] = (
    product_revenue['Sales'].cumsum()
)

# 4. Total Sales
total_sales = product_revenue['Sales'].sum()

# 5. Calculate Cumulative Revenue %
product_revenue['Cumulative Revenue %'] = (
    product_revenue['Cumulative Sales']
    / total_sales
    * 100
).round(2)

# 6. Products Contributing 80% Revenue
top_products = product_revenue[
    product_revenue['Cumulative Revenue %'] <= 80
]

# 7. Percentage of Products Contributing 80% Revenue
product_percentage = (
    len(top_products)
    / len(product_revenue)
    * 100
)

# 8. Display Results
print("Total Sales:", total_sales)
print("\nProducts Contributing 80% Revenue:")
print(top_products)

print("\nPercentage of Products Contributing 80% Revenue:")
print(round(product_percentage, 2), "%")

# Complete Table
print("\nComplete Pareto Analysis:")
product_revenue.reset_index()
```

Key Performance Indicators (KPIs)
KPI	Description
Gross Margin (%)	Gross Profit ÷ Sales
Profit per Unit	Gross Profit ÷ Units
Revenue Contribution	Product sales ÷ total sales
Profit Contribution	Product profit ÷ total profit
Margin Volatility	Variability of margin over time
